A repo whose name equals my username becomes the profile page. A repo marked as a **template** can create new repos with the same files.

## Commands (what each one does)
| Command | Meaning |
|---|---|
| `gh auth login` | Log the GitHub CLI in (browser flow) |
| `gh repo create NAME --public --source=. --push` | Create a repo from a folder that is already a Git repo and upload it |
| `gh repo edit OWNER/REPO --template` | Mark a repo as a template |
| `gh repo create NAME --template BuiltBySwap/repo-template --public --clone` | Create a new repo from the template and clone it (already a Git repo, no `git init` needed) |

## Rules for every repo
Problem section first in the README, a Mermaid architecture diagram, small conventional commits (`feat:`, `fix:`, `docs:`), CI from the first week, numbers in the README, no secrets, no employer code.

## Bug diary

| Error or symptom | Cause | Fix |
|---|---|---|
| `gh repo create --source=.` failed | The folder was not a Git repo yet | `git init`, `git add .`, `git commit`, then create |
