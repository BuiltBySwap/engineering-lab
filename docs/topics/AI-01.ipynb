## Goal
Know what an LLM API call really is, and what the bill is made of.

## What I learned (plain words)
- A **model** guesses the next small piece of text (a **token**), one at a time.
- The API is **stateless**. The app resends the whole conversation each time, so long chats cost more.
- The **context window** is how much the model can read in one request.
- **Temperature** controls how adventurous the guesses are. Low temperature gives predictable output (good for JSON that code parses). It does not change the cost.
- You pay for **input + output + hidden thinking tokens**.
- An API key must never be inside an APK. Put it on a server.

## Examples: the same task, three prices
| Version | Thinking | Output | Total |
|---|---|---|---|
| Default | 215 | 58 | 283 |
| Low thinking | 0 | 119 | 129 |
| Low thinking + exact instruction + output cap | 0 | 8 | 25 |

## Bugs and fixes
| Error or symptom | Cause | Fix |
|---|---|---|
| 404 | Model name wrong, or the model was retired for new users | List models, then use a current one |
| 503 | The provider is overloaded (their side) | Retry with backoff, or fall back to another model |
| 429 | I sent too many requests (my side) | Slow down, back off, queue |
| `Thinking level MINIMAL is not supported` | Features differ per model | Use `low` for that model |
| `answer: null` from my jq filter | The API returned an error and my filter hid it | Print `.error` too |
