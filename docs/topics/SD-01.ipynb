## Practice: design a weather app
Interviewer's answers to my clarifying questions: severe alerts reach phones within about 60 seconds, even when the app is closed; background refresh about every 3 hours; data under 15 minutes old is fine on screen; v1 features are current conditions, 24-hour hourly, 7-day forecast, up to 10 saved cities, current location; one widget; tablets and foldables work; dark theme follows the system; offline shows the last data with a "last updated" time; 5 million daily users; data comes from a third-party provider with rate limits and a cost per call; 20 languages and °C/°F.

Questions I missed: offline behaviour, location, scale.

## AI feature: severe-alert simplifier (server)
```text
Alert feed → ingest (dedupe) → match areas ─┬─ FAST path: push the OFFICIAL text now
                                            └─ AI path: simplify × language → check → cache → follow-up
```
- Cost: one AI call per alert × language, not per user.
- Safety: always show the official text next to the AI version, and check that times, numbers and places match the source.
- Failure: if the AI is slow or down, the alert still goes out. Keep slow, optional work off the critical path.

## Quiz and mistakes
- The push must not wait for the AI.
