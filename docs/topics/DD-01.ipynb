## How the site is built (commands explained)

I write Markdown and notebooks. MkDocs turns them into HTML. GitHub Actions builds on every push. GitHub Pages hosts the result for free.

```text
docs/*.md, *.ipynb ──git push──▶ GitHub Actions: install Python + MkDocs → mkdocs build → upload site/ ──▶ GitHub Pages
```

| Command | What it does |
|---|---|
| `gh repo create engineering-lab --template BuiltBySwap/repo-template --public --clone` | Creates the repo on GitHub from my template and clones it |
| `cp -R src/. dest/` | Copies files, including hidden folders like `.github` |
| `gh api -X POST repos/BuiltBySwap/engineering-lab/pages -f build_type=workflow` | Turns on GitHub Pages, built by Actions |
| `git add . && git commit -m "..." && git push` | Saves a snapshot and uploads it. The push triggers the build |

- **The command never builds the site.** The push triggers `.github/workflows/pages.yml`, and a robot on GitHub runs `mkdocs build --strict`.
- **The domain is free** because `github.io` belongs to GitHub. Every user gets a subdomain: `builtbyswap.github.io/engineering-lab/`. GitHub finds my files from the `Host` header.
- **The site is static** (pre-built HTML), so there is no server, database or hosting bill. Compare with a Ktor server, which runs code on every request.

## Bugs and fixes
| Error or symptom | Cause | Fix |
|---|---|---|
| Pasted command text ended up inside `settings.gradle.kts` | A multi-line shell block was pasted into a file | Write files from the editor, or paste into the terminal prompt |
| `./gradlew: no such file or directory` | Terminal was in the wrong folder | `cd` into the project |
| Tests failed to compile | Test file was in `src/main` | Move it to `src/test` |
