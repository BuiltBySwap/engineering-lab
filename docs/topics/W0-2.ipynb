## 1 · Variables: `val` / `var` → just assign

````python
name = "Swapnil"          # no type needed
count = 1
count += 1
MAX_RETRIES = 3           # UPPERCASE = "please don't change me" (a convention, NOT enforced)
print(name, count, MAX_RETRIES)
````

## 2 · Collections: `listOf` / `mapOf` → list `[]`, dict `{}`

````python
numbers = [1, 2, 3]                      # mutableListOf(1, 2, 3)
ages = {"asha": 30, "ravi": 25}          # mutableMapOf("asha" to 30, ...)
ages["meera"] = 28
print(numbers, ages, ages.get("zoya", 0))   # get with a default, like getOrDefault
````

## 3 · Comprehensions: `filter { }.map { }` → `[x for x in xs if cond]`

````python
squares = [n * n for n in numbers if n > 1]    # numbers.filter { it > 1 }.map { it * it }
print(squares)
````

## 4 · Functions: `fun` → `def`, with a colon, indentation and f-strings

````python
def greet(name, excited=False):          # fun greet(name: String, excited: Boolean = false)
    suffix = "!" if excited else "."
    return f"Hello, {name}{suffix}"      # string template "Hello, $name"

print(greet("Asha"), greet("Ravi", excited=True))
````

## 5 · Classes: `data class` → `@dataclass`

````python
from dataclasses import dataclass

@dataclass
class PlanDay:                 # data class PlanDay(var id: String, var hours: Double)
    id: str
    hours: float

day = PlanDay("BE-01", 1.17)
print(day, day == PlanDay("BE-01", 1.17))
````

## 6 · Null safety: `null` → `None`

- `x ?: default` is closest to `x if x is not None else default` (safe). `x or default` is shorter but also replaces `0`, `""` and `[]`.
- Always write `is None`, never `== None`. Python has no `?.`, so check explicitly.
- `a if cond else b` is Kotlin's `if/else` *expression*, **not** the Elvis operator.

````python
nickname = None
display = nickname if nickname is not None else "anonymous"
print(display)
````

## 7 · Exceptions: `try/catch` → `try/except`

````python
try:
    int("abc")
except ValueError as e:                 # catch (e: NumberFormatException)
    print("bad number:", e)
finally:
    print("always runs")
````

## 8 · Modules: `package` / `import` → `from module import name`

````python
from math import sqrt        # import kotlin.math.sqrt
import math as m             # import ... as m
print(sqrt(16), m.pi)
````

## 9 · Tuples: `Pair` / `Triple` → `(a, b)` and unpacking

````python
point = (3, 4)
x, y = point                 # val (x, y) = point
print(x, y)
````

## 10 · Loops: `0 until n` / `withIndex()` → `range()` / `enumerate()`

````python
for i in range(3):
    print(i, end=" ")
print()
for index, value in enumerate(["a", "b"]):   # for ((index, value) in list.withIndex())
    print(index, value)
````

## 11 · NumPy basics: arrays, shape, dot product, broadcasting, boolean mask

````python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape, a.reshape(3, 2).shape)       # (2, 3) (3, 2)
print(np.dot([1, 2, 3], [4, 5, 6]))         # 32
print(a + 10)                               # broadcasting: 10 is added to every element
print(a[a > 3])                             # boolean mask -> [4 5 6]
````

## Three corrections to my own first recap
1. **Constants:** Python has no `val`. `MAX_VAL` in uppercase is only a convention. `from typing import Final; MAX_VAL: Final = 10` is a hint for tools, still not enforced at run time.
2. **Null safety:** `a if cond else b` is the `if/else` expression, not `?:` (see section 6).
3. **Data classes:** `@dataclass` fields are changeable like `var`. For Kotlin-style `val`, use `@dataclass(frozen=True)`.

````python
from dataclasses import dataclass

@dataclass(frozen=True)
class Point:
    x: int
    y: int

p = Point(1, 2)
try:
    p.x = 5
except Exception as e:
    print(type(e).__name__, "- frozen works like val")
````

## Five gotchas that cause most beginner bugs
**Predict each output before you run the cell.**

````python
# 1. Lists are shared references (no copy on assignment)
a = [1, 2]
b = a
b.append(3)
print(a)                 # ?
c = a.copy()
c.append(4)
print(a, c)              # ?
````

````python
# 2. The mutable default argument trap
def add_item(x, items=[]):
    items.append(x)
    return items

print(add_item(1), add_item(2))     # ?  (one list is created once and reused)

def add_item_fixed(x, items=None):
    if items is None:
        items = []
    items.append(x)
    return items

print(add_item_fixed(1), add_item_fixed(2))
````

````python
# 3. Division, and 4. slicing
print(7 / 2, 7 // 2)                  # ?
nums = [10, 20, 30, 40, 50]
print(nums[1:4], nums[::-1], nums[-1])    # ?
````

````python
# 5. NumPy: axis, and views vs copies
m = np.array([[1, 2, 3], [4, 5, 6]])
print(m.sum(axis=0), m.sum(axis=1))   # ?  (axis=0 sums down the columns, axis=1 across the rows)
v = m[0, :2]
v[0] = 99
print(m)                              # ?  v is a VIEW of m
w = m[0, :2].copy()
w[0] = -1
print(m)                              # unchanged: a copy
````
